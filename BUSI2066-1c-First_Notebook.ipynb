## 2 · Store a value in a variable

A **variable** stores a value under a name, using `=`. After that, the name stands for the value.

````python
ticket_price = 1280    # HK$ per ticket, standing zone
tickets = 4

ticket_price * tickets
````

A cell shows only its **last** line. To see more than one thing, use `print()`.

````python
total = ticket_price * tickets + 90    # + HK$90 booking fee

print(total)
print(total / tickets)                 # average cost per ticket
````
